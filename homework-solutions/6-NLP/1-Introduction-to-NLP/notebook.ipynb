# Introduction to NLP: Marvin and an ELIZA experiment

This notebook executes the basic bot exercise, then investigates the published
[NLTK ELIZA implementation](https://www.nltk.org/_modules/nltk/chat/eliza.html)
for the **Search for a bot** assignment. The transcript is an actual local run,
with a fixed random seed. No interview or test on another person is claimed.
See [answers.md](answers.md) for the paper and memory challenge.

```python
import random
from pathlib import Path
import nltk
import pandas as pd
from IPython.display import display
from nltk.chat.eliza import pairs, reflections
from nltk.chat.util import Chat
from bot import reply, RESPONSES

nltk.data.path.insert(0, str(Path('../data/nltk_data').resolve()))
random.seed(42)
messages = ['I am good thanks', 'Today I went for a walk', 'My team lost', ' BYE ']
transcript = pd.DataFrame({'input': messages, 'response': [reply(text) for text in messages]})
assert transcript.response.iloc[-1] is None
assert all(response in RESPONSES for response in transcript.response.iloc[:-1])
display(transcript)
```

## Assignment: controlled ELIZA probes

The prompts test reflection, nonsensical assertions, negation, recall, and an
unrelated factual question. `Chat.respond()` uses the library's actual rules.

```python
random.seed(42)
eliza = Chat(pairs, reflections)
probes = [
    'I am sad.',
    'You are a bicycle.',
    'I am not sad.',
    'My brother broke my bicycle.',
    'What did I say about my brother?',
    'Why is the sky blue?',
]
observations = pd.DataFrame({'input': probes, 'response': [eliza.respond(text) for text in probes]})
assert observations.response.notna().all()
display(observations)
observations.to_csv('eliza-transcript.csv', index=False)
```

## What the experiment establishes

The bicycle prompt elicits a question accepting the bicycle premise. The recall
prompt gets a generic question, rather than the earlier event. Negation is
copied into a template without an independent check of what was negated.
These failures are consistent with the inspected regular-expression rules;
they are not evidence about every bot or about the historical ELIZA program.

Marvin's random reply also does not depend on the message's meaning. The
challenge design stores a topic and sentiment between turns; the next
lesson's [bot](../2-Tasks/bot.py) implements a small version of that idea.
